```python
import pandas as pd
import numpy as np
```

```python

df = pd.read_csv('upi_fake_data.csv')
```

```python

# 1. Duplicate check - Check If fake data is there or not
df = df.drop_duplicates(subset=['Transaction_ID'])
```

```python

# 2. Missing values check
print(df.isnull().sum())  
```

```python

# 3. Data Type fix  - Important to Power BI 
df['Time'] = pd.to_datetime(df['Time'])
df['Amount'] = df['Amount'].astype(float)
df['Is_Fraud'] = df['Is_Fraud'].astype(int)
```

```python

# 4. New columns for analysis 
df['Hour'] = df['Time'].dt.hour
df['Day'] = df['Time'].dt.day_name()
df['Date'] = df['Time'].dt.date
```

```python

# 5. Outlier handle - Amount 10,000+ is error
df = df[df['Amount'] < 10000] 
```

```python

df.to_csv('upi_cleaned.csv', index=False)
print("✅ Cleaning Done. Rows:", len(df))
```
